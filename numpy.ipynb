```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from  sklearn.model_selection import train_test_split
```

```python
#creat database
np.random.seed(42)
```

```python
df=pd.DataFrame({
    'Student_id': range(1,101),
    'Name':[f'student_{i}'for i in range(1,101)],#use the for loop
    'Gender':np.random.choice(['male','female'],100),
    'Department':np.random.choice(['CSE','ECE','ME','CE'],100),
    'Mark':np.random.randint(40,101,100)
})
```

```python
df
```

```python
df.head(100)
```

```python
print("population size",len(df))
```

```python
#about data/popution\
print("department",df['Department'].value_counts())
print ("gender",df['Gender'].value_counts())
print ("basic distrute",df['Mark'].describe())
```

```python
# simple random Sampling
sample_size=10
simple_random_sample=df.sample(
  n=sample_size,
  random_state=42)
```

```python

simple_random_sample
```

```python
# Visualization for simple random sample
plt.figure(figsize=(12,6))

# Population scatter
plt.scatter(df['Student_id'], df['Mark'], alpha=0.35, label='Population')

# Selected sample scatter
plt.scatter(simple_random_sample['Student_id'],
            simple_random_sample['Mark'],
            s=120, label='Selected Sample')

# Labels and title
plt.title('Simple Random Sample')
plt.xlabel('Student_id')
plt.ylabel('Mark')
plt.legend()
plt.show()
```

```python
N=len(df)
n=10
k=N//n
start =np.random.randint(0,k)
systematic_sampling = df.iloc[start::k].head(n)
print("Sampling Inderval :",k)
print("start pos:",start)
systematic_sampling
```

```python
plt.figure(figsize=(12,6))

# Population scatter
plt.scatter(df['Student_id'], df['Mark'], alpha=0.35, label='Population')

# Selected sample scatter
plt.scatter(systematic_sampling['Student_id'],systematic_sampling['Mark'],
            s=120,label='Selected Sample')

# Labels and title
plt.xlabel('Student_id')
plt.ylabel('Mark')
plt.title('Systematic Sample')
plt.legend()
plt.show()
```

visulization of sysematic sample

```python
Sample_size=20
stratified_sample , remaining = train_test_split(
    df,
    test_size=1-Sample_size/len(df),
    stratify=df['Department'],
    random_state=42
)
stratified_sample
```

```python
# cluster sampling
cluster= df ['Department'].unique()
selecter_cluster=np.random.choice(
    cluster,
    size=2,
    replace=False)
cluster_sample=df[
    df['Department'].isin(selecter_cluster)]
print("select cluster:",selecter_cluster)
print ("\n sample_size:",len(cluster_sample))
print (cluster_sample)
```

```python
#quota sampling
male_sample=df[
    df['Gender']=='Male'
    ].sample(n=5,random_state=42)

femal_sample=df[
    df['Gender']=='Female'
    ].sample(n=5, random_state=42)

quota_sample=pd.concat([male_sample,femal_sample])
quota_sample
```
